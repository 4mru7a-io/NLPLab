```python
import spacy
from spacy import displacy
nlp = spacy.load("en_core_web_sm")
```

```python
txt = "the cat is sleeping on the sofa"
doc = nlp(txt)
```

```python
print(f"{'token':<13} | {'fine pos tag':<13} | {'coarse pos':<13} | {'explanation':<20}")

for token in doc:
    explanation = spacy.explain(token.tag_) or ""
    print(f"{token.text:<13} | {token.tag_:<13} | {token.pos_:<13} | {explanation:<20}")
```

```python
print(f"{"token.text": <13} | {"relation" : <13} | {"head pos" :<13} | {"head word" : <13}")
```

```python
for token in doc:
    print(f"{token.text:<13} | {token.dep_:<13} | {token.head.pos_:<13} | {token.head.text:<13}")
```

```python
displacy.serve(doc, style="dep")
```
