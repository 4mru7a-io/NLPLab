```
import re
text="My name is Balaji Mukkawar. My date of birth is 11-02-2005. my contact number is 8624910075 and email is balajimukkawar2005@gmail.com"
print("Email",re.findall(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b',text))
print("Mobile",re.findall(r'\b[6-9]\d{9}\b',text))
print("Date",re.findall(r'\b\d{2}-\d{2}-\d{4}\b',text))
print("Name",re.findall(r'\b[A-Z][a-z]+\b',text))
```

```
def contains_abba_fsa(word):
    state = 0
    for char in word.lower():
        if state == 0:
            state = 1 if char == 'a' else 0
        elif state == 1:
            state = 2 if char == 'b' else (1 if char == 'a' else 0)
        elif state == 2:
            state = 3 if char == 'b' else (1 if char == 'a' else 0)
        elif state == 3:
            state = 4 if char == 'a' else 0
        elif state == 4:
            break
    return state == 4
sample_text = "We saw a cabbage, an abbot, and some crabmeat in the cellar of abba."
words = sample_text.split()

matched_words = [word.strip(".,!;")for word in words if contains_abba_fsa(word.strip(".,!;"))]
print("Words containing 'abba':", matched_words)
```
