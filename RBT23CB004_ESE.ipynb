```python
#Write a program to perform basic NLP preprocessing t5asks such as tokenization,stop-word removal,stemming and
#lemmatization on a given English text.
#Demonstrate the challenges involved in processing Indian language text using a sample dataset.


import nltk
import re

# Download required NLTK datasets
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')
nltk.download('omw-1.4')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Sample English text
text = """Natural Language Processing is an interesting field.
It helps computers understand human languages.
Students are learning different NLP techniques."""

print("Original Text:")
print(text)

# 1. Tokenization
tokens = word_tokenize(text.lower())
print("\n1. Tokenization:")
print(tokens)

#2.Stopword Removal
stop_words = set(stopwords.words('english'))

filtered = []
for word in tokens:
    if word.lower() not in stop_words and word.isalpha():
        filtered.append(word)
print("\n2.After Stopword Removal:", filtered)

# 3. Stemming
stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word) for word in filtered_words
]

print("\n3. Stemming:")
print(stemmed_words)

# 4. Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word, pos='v')
    for word in filtered_words
]

print("\n4. Lemmatization:")
print(lemmatized_words)
```
