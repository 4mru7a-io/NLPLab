```python
import re
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
```

```python
pages = [
    "Student name roll number college university marks",
    "Subject mathematics physics chemistry examination",
    "Student details admission course department",
    "Invoice number customer amount total payment",
    "Product price quantity bill tax amount",
    "Customer name invoice date payment details"
]

labels = [
    "Document_A",
    "Document_A",
    "Document_A",
    "Document_B",
    "Document_B",
    "Document_B"
]
```

```python
vectorizer = CountVectorizer()

X = vectorizer.fit_transform(pages)

model = MultinomialNB()

model.fit(X, labels)

predictions = model.predict(X)

print("Page Classification:")

for i, pred in enumerate(predictions):
    print("Page", i+1, ":", pred)

# Probabilistic Classification using Naive Bayes
```

```python
documents = []
current_document = []
current_state = None

for i in range(len(pages)):

    predicted_state = predictions[i]

    if current_state is None:
        current_state = predicted_state

    if predicted_state != current_state:

        documents.append(current_document)

        current_document = []

        current_state = predicted_state

    current_document.append(pages[i])

documents.append(current_document)
# Finite State Machine for Document Separation(state A- doc A, state B- doc B)
#state changes new doc starts
```

```python
for i, doc in enumerate(documents):

    print("\nDocument", i+1)

    for page in doc:
        print(page)
# to Display Separated Documents
```

```python
print("Input Page Sequence:")
print(predictions)

print("\nSeparated Document Sequence:")

for i, doc in enumerate(documents):
    print("Document", i+1, "contains", len(doc), "pages")
# Sequence Mapping Output
```

```python
for i, doc in enumerate(documents):

    print("\n========== DOCUMENT", i+1, "==========")

    for j, page in enumerate(doc):
        print("Page", j+1, ":", page)
# actual sep docs content
```
