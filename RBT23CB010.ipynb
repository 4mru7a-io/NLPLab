```python
import nltk
import re

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger_eng')

text = """The students are studying NLP.
They are playing games and running quickly."""

tokens = word_tokenize(text)
print("Tokens:", tokens)

stop_words = set(stopwords.words('english'))
filtered_words = [
    word for word in tokens
    if word.lower() not in stop_words
    and re.search(r'[a-zA-Z]', word)
]
print("After Stop Word Removal:", filtered_words)
stemmer = PorterStemmer()
stemmed_words = [
    stemmer.stem(word) for word in filtered_words
]
print("After Stemming:", stemmed_words)

lemmatizer = WordNetLemmatizer()
lemmatized_words = [
    lemmatizer.lemmatize(word, pos='v')
    for word in filtered_words
]
print("After Lemmatization:", lemmatized_words)

indian_text = "मुझे हिंदी आती है"
print("Indian Language Text:", indian_text)
print("Indian Language Tokens:", indian_text.split())

```
