```python
import re

text = """Ravi went to the library.
He borrowed a book.
The book was interesting.
So, he started reading it at home. He didn't find the time to read it"""
```

```python
sentences = re.split(r'(?<=[.!?])\s+', text)

print("SENTENCES:")
for i, s in enumerate(sentences, 1):
    print(i, s)
```

```python
entities = []
for word in re.findall(r'\b[A-Za-z]+\b', text):
  if word[0].isupper():
    entities.append(word)
print(entities)
```

```python
previous = []

for i, sentence in enumerate(sentences, 1):
    words = sentence.split()

    for word in words:
        if word.lower() in ["he", "she", "his", "her"]:
            if previous:
                print(word, "->", previous[-1])

    for word in words:
        if word[0].isupper() and word.lower() not in ["the", "so"]:
            previous.append(word)
```

```python
markers = ["because", "therefore", "however", "so", "then", "also"]

for i in range(len(sentences) - 1):
    words = sentences[i + 1].lower().split()

    common = set(sentences[i].lower().split()) & set(words)
    marker = [m for m in markers if m in words]

    print("\nSentence", i+1, "->", i+2)
    print("Shared words:", common)
    print("Discourse marker:", marker)
```
