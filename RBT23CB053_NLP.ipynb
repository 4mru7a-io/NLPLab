```python
import nltk
import spacy
from nltk import word_tokenize,pos_tag,tree

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger_eng')
nlp = spacy.load('en_core_web_sm')
sentence = "The boy is playing cricket"

words = word_tokenize(sentence)
print('POS Tags:')
print(pos_tag(words))

doc = nlp(sentence)
print('\nDependency Tree:')
for token in doc:
    print(token.text, "-->", token.dep_, "-->", token.head.text)

Grammer = nltk.CFG.fromstring("""
  S -> NP VP
  NP -> Det N | N
  VP -> Aux V NP | V NP
  Det -> 'The' | 'the'
  N -> 'boy' | 'cricket'
  Aux -> 'is'
  V -> 'playing'
""")

parser = nltk.ChartParser(Grammer)
print("\nParsing Tree:")
for tree_parsed in parser.parse(words):
    print(tree_parsed)
```
